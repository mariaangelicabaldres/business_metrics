### Clean Raw Data - Remove Blank Emails

```python
# Create a working copy
df_clean = df_raw.copy()

print(f"Initial dataset: {len(df_clean)} rows")

# Check for blank/null emails
blank_emails = df_clean['Email'].isna() | (df_clean['Email'].astype(str).str.strip() == '')
blank_count = blank_emails.sum()

if blank_count > 0:
    print(f"\n⚠️ Found {blank_count} rows with blank Email")
    print("Rows to be removed (showing key columns only):")
    
    # Display only key columns to avoid wide output
    key_columns = ['Start Date', 'Response Type', 'Email']
    available_key_columns = [col for col in key_columns if col in df_clean.columns]
    
    if available_key_columns:
        display(df_clean[blank_emails][available_key_columns].head(10))
    else:
        display(df_clean[blank_emails].iloc[:, :5].head(10))
    
    # Remove rows with blank emails
    df_clean = df_clean[~blank_emails].copy()
    print(f"✓ Removed {blank_count} rows with blank Email")
else:
    print("\n✓ No blank emails found")

print(f"\nDataset after removing blank emails: {len(df_clean)} rows")
```

### Clean Duplicates Based on Email - Step 1 (Identify Discovery/Intro Call Types)

```python
# Identify which rows are for Discovery Calls and which are for Intro Calls
# Based on whether there's a link in the respective columns

# CRITICAL: Normalize email to lowercase FIRST
df_clean['Email'] = df_clean['Email'].str.lower().str.strip()

# First, identify the Start Date column
start_date_col = None
for col in df_clean.columns:
    if 'start' in col.lower() and 'date' in col.lower():
        start_date_col = col
        break

if start_date_col is None:
    for col in df_clean.columns:
        if 'date' in col.lower() and 'utc' in col.lower():
            start_date_col = col
            break

print(f"✓ Start Date column identified: '{start_date_col}'")

# Check column names for discovery and intro call columns
discovery_col = None
intro_col = None

for col in df_clean.columns:
    if 'discovery' in col.lower() and 'book' in col.lower():
        discovery_col = col
    if 'intro' in col.lower() and 'book' in col.lower():
        intro_col = col

print(f"Discovery Call Column: {discovery_col}")
print(f"Intro Call Column: {intro_col}")

# Add helper columns to identify call type
if discovery_col and intro_col:
    # A row is for Discovery Call if there's a non-empty value in the discovery column
    df_clean['Is_Discovery_Call'] = df_clean[discovery_col].notna() & (df_clean[discovery_col].astype(str).str.strip() != '')
    
    # A row is for Intro Call if there's a non-empty value in the intro column
    df_clean['Is_Intro_Call'] = df_clean[intro_col].notna() & (df_clean[intro_col].astype(str).str.strip() != '')
    
    print("\n✓ Call type identification complete")
    print(f"Discovery Calls: {df_clean['Is_Discovery_Call'].sum()}")
    print(f"Intro Calls: {df_clean['Is_Intro_Call'].sum()}")
else:
    print("\n⚠️ Warning: Could not find discovery/intro call columns")
    df_clean['Is_Discovery_Call'] = False
    df_clean['Is_Intro_Call'] = False
```

### Clean Duplicates Based on Email - Step 2 (Process Duplicates)

```python
# Re-identify the Start Date column if not already in memory
try:
    start_date_col
    print(f"✓ Using existing start_date_col: {start_date_col}")
except NameError:
    # Find the Start Date column
    start_date_col = None
    print(f"Available columns: {df_clean.columns.tolist()}")
    
    # First try to find "Start Date"
    for col in df_clean.columns:
        if 'start' in col.lower() and 'date' in col.lower():
            start_date_col = col
            break
    
    # If not found, look for any column with "date" and "utc"
    if start_date_col is None:
        for col in df_clean.columns:
            if 'date' in col.lower() and 'utc' in col.lower():
                start_date_col = col
                break
    
    print(f"✓ Identified Start Date column: {start_date_col}")

if start_date_col is None:
    raise ValueError("Could not find Start Date column!")

# Convert Start Date to datetime for comparison
df_clean[start_date_col] = pd.to_datetime(df_clean[start_date_col], format='mixed', errors='coerce')

# Sort by Email, Response Type (completed comes before partial alphabetically), and Start Date (latest first)
df_clean = df_clean.sort_values(
    by=['Email', 'Response Type', start_date_col],
    ascending=[True, True, False]
).reset_index(drop=True)

# Group by Email to process duplicates
email_groups = df_clean.groupby('Email')

rows_to_keep = []
duplicate_removal_log = []

for email, group in email_groups:
    if len(group) == 1:
        # Only one entry for this email, keep it
        rows_to_keep.append(group.index[0])
    else:
        # Multiple entries for this email
        partials = group[group['Response Type'] == 'partial']
        completed = group[group['Response Type'] == 'completed']
        
        # Rule 1: If we have ANY completed, remove ALL partials for this email
        if len(completed) > 0:
            # Log removal of partials
            if len(partials) > 0:
                duplicate_removal_log.append({
                    'Email': email,
                    'Rule': 'Remove ALL Partials (Completed exists)',
                    'Removed Partials': len(partials),
                    'Kept Completed': len(completed)
                })
            
            # Now process the completed entries
            if len(completed) == 1:
                # Only 1 completed, keep it
                rows_to_keep.append(completed.index[0])
            elif len(completed) >= 2:
                # Multiple completed - check call types
                discovery_completed = completed[completed['Is_Discovery_Call'] == True]
                intro_completed = completed[completed['Is_Intro_Call'] == True]
                
                # If we have both discovery and intro, keep one of each (latest of each type)
                if len(discovery_completed) > 0 and len(intro_completed) > 0:
                    # Keep the latest discovery call
                    latest_discovery = discovery_completed.iloc[0]
                    rows_to_keep.append(latest_discovery.name)
                    
                    # Keep the latest intro call
                    latest_intro = intro_completed.iloc[0]
                    rows_to_keep.append(latest_intro.name)
                    
                    duplicate_removal_log.append({
                        'Email': email,
                        'Rule': 'Keep both Discovery and Intro Completed',
                        'Removed Partials': len(partials),
                        'Kept Completed': 2
                    })
                else:
                    # All completed are same type, keep only the latest
                    latest_completed = completed.iloc[0]
                    duplicate_removal_log.append({
                        'Email': email,
                        'Rule': 'Keep latest Completed (same type)',
                        'Removed Partials': len(partials),
                        'Kept Completed': 1
                    })
                    rows_to_keep.append(latest_completed.name)
        
        # Rule 2: If we have ONLY partials (no completed), keep the latest partial
        else:
            if len(partials) >= 2:
                latest_partial = partials.iloc[0]  # Already sorted by date descending
                duplicate_removal_log.append({
                    'Email': email,
                    'Rule': 'Keep latest Partial (no Completed)',
                    'Removed Partials': len(partials) - 1,
                    'Kept Completed': 0
                })
                rows_to_keep.append(latest_partial.name)
            elif len(partials) == 1:
                # Only 1 partial, keep it
                rows_to_keep.append(partials.index[0])

# Create cleaned dataframe
df_cleaned = df_clean.loc[rows_to_keep].copy()

# Display duplicate removal summary
print("\n" + "=" * 60)
print("DUPLICATE REMOVAL SUMMARY")
print("=" * 60)
print(f"Rows before duplicate removal: {len(df_clean)}")
print(f"Rows after duplicate removal: {len(df_cleaned)}")
print(f"Total rows removed: {len(df_clean) - len(df_cleaned)}")

if duplicate_removal_log:
    print("\nDuplicate Removal Log (first 20 entries):")
    df_log = pd.DataFrame(duplicate_removal_log)
    display(df_log.head(20))
    print(f"\nTotal duplicate groups processed: {len(df_log)}")
else:
    print("\n✓ No duplicates found")

# Verify Response Type counts
print("\nResponse Type counts in cleaned data:")
print(df_cleaned['Response Type'].value_counts())

print("=" * 60)
```

### Create Daily Aggregated Data

```python
# Work with cleaned data
df_daily_prep = df_cleaned.copy()

print(f"✓ Using '{start_date_col}' column for date aggregation")
print(f"Total rows in cleaned data: {len(df_daily_prep)}")

# Convert Start Date to date only (extract date from the UTC timestamp)
df_daily_prep['AggregationDate'] = pd.to_datetime(df_daily_prep[start_date_col]).dt.date

# Identify discovery and intro call columns
discovery_col = None
intro_col = None

for col in df_daily_prep.columns:
    if 'discovery' in col.lower() and 'book' in col.lower():
        discovery_col = col
        print(f"✓ Found Discovery Call Column: {discovery_col}")
    if 'intro' in col.lower() and 'book' in col.lower():
        intro_col = col
        print(f"✓ Found Intro Call Column: {intro_col}")

# Group by AggregationDate and count each response type separately
df_daily_counts = df_daily_prep.groupby(['AggregationDate', 'Response Type']).size().unstack(fill_value=0).reset_index()

# Rename columns (response types are lowercase in the data)
if 'partial' in df_daily_counts.columns:
    df_daily_counts.rename(columns={'partial': 'Partials'}, inplace=True)
else:
    df_daily_counts['Partials'] = 0

if 'completed' in df_daily_counts.columns:
    df_daily_counts.rename(columns={'completed': 'Completed'}, inplace=True)
else:
    df_daily_counts['Completed'] = 0

# Aggregate Discovery and Intro Calls by date (count non-empty values)
if discovery_col:
    discovery_counts = df_daily_prep[df_daily_prep[discovery_col].notna() & (df_daily_prep[discovery_col].astype(str).str.strip() != '')].groupby('AggregationDate').size().reset_index(name='Discovery Calls')
else:
    discovery_counts = None

if intro_col:
    intro_counts = df_daily_prep[df_daily_prep[intro_col].notna() & (df_daily_prep[intro_col].astype(str).str.strip() != '')].groupby('AggregationDate').size().reset_index(name='Intro Calls')
else:
    intro_counts = None

# Merge all counts
df_daily = df_daily_counts[['AggregationDate', 'Partials', 'Completed']].copy()

if discovery_counts is not None:
    df_daily = df_daily.merge(discovery_counts, on='AggregationDate', how='left')
    df_daily['Discovery Calls'] = df_daily['Discovery Calls'].fillna(0).astype(int)
else:
    df_daily['Discovery Calls'] = 0

if intro_counts is not None:
    df_daily = df_daily.merge(intro_counts, on='AggregationDate', how='left')
    df_daily['Intro Calls'] = df_daily['Intro Calls'].fillna(0).astype(int)
else:
    df_daily['Intro Calls'] = 0

# Rename AggregationDate to Date
df_daily.rename(columns={'AggregationDate': 'Date'}, inplace=True)

# Reorder columns
df_daily = df_daily[['Date', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']]

# Sort by date
df_daily = df_daily.sort_values('Date').reset_index(drop=True)

# Convert Date to datetime then format as MM/DD/YYYY
df_daily['Date'] = pd.to_datetime(df_daily['Date']).dt.strftime('%m/%d/%Y')

print("\n✓ Daily data aggregated successfully")
print(f"\nDaily dataset shape: {df_daily.shape[0]} rows × {df_daily.shape[1]} columns")
print(f"\nFirst 5 days:")
display(df_daily.head(5))

print(f"\nLast 5 days:")
display(df_daily.tail(5))

# Verification summary
print(f"\n" + "=" * 60)
print("DAILY AGGREGATION VERIFICATION")
print("=" * 60)
print(f"Total Partials: {df_daily['Partials'].sum()}")
print(f"Total Completed: {df_daily['Completed'].sum()}")
print(f"Total Discovery Calls: {df_daily['Discovery Calls'].sum()}")
print(f"Total Intro Calls: {df_daily['Intro Calls'].sum()}")
print(f"Total Submissions: {df_daily['Partials'].sum() + df_daily['Completed'].sum()}")
print("=" * 60)
```

### Create Weekly Aggregated Data

```python
# Work with daily data
df_weekly_prep = df_daily.copy()

# Convert Date to datetime
df_weekly_prep['Date'] = pd.to_datetime(df_weekly_prep['Date'])

# Create week identifier (week starting on Sunday)
# Sunday = 6 in dayofweek, we want Sunday to map to the start of its week
# Formula: subtract (dayofweek + 1) % 7 days
df_weekly_prep['Week'] = df_weekly_prep['Date'] - pd.to_timedelta((df_weekly_prep['Date'].dt.dayofweek + 1) % 7, unit='D')

# Aggregate by week
df_weekly = df_weekly_prep.groupby('Week').agg({
    'Date': ['min', 'max'],
    'Partials': 'sum',
    'Completed': 'sum',
    'Discovery Calls': 'sum',
    'Intro Calls': 'sum'
}).reset_index()

# Flatten column names
df_weekly.columns = ['Week', 'Week_Start', 'Week_End', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']

# Create Week Number (sequential)
df_weekly['Week_Num'] = range(1, len(df_weekly) + 1)

# Format date range
df_weekly['Date Range'] = (df_weekly['Week_Start'].dt.strftime('%m/%d/%Y') + ' - ' + 
                           df_weekly['Week_End'].dt.strftime('%m/%d/%Y'))

# Select and reorder columns
df_weekly = df_weekly[['Week_Num', 'Date Range', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']]

# Rename Week_Num to Week
df_weekly.columns = ['Week', 'Date Range', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']

print("✓ Weekly data aggregated successfully")
print(f"\nWeekly dataset shape: {df_weekly.shape[0]} rows × {df_weekly.shape[1]} columns")
print(f"\nFirst 5 weeks:")
display(df_weekly.head(5))

print(f"\nLast 5 weeks:")
display(df_weekly.tail(5))
```

### Create Monthly Aggregated Data

```python
# Work with daily data
df_monthly_prep = df_daily.copy()

# Convert Date to datetime
df_monthly_prep['Date'] = pd.to_datetime(df_monthly_prep['Date'])

# Extract year and month
df_monthly_prep['Year'] = df_monthly_prep['Date'].dt.year
df_monthly_prep['Month_Num'] = df_monthly_prep['Date'].dt.month
df_monthly_prep['Month_Name'] = df_monthly_prep['Date'].dt.strftime('%B')

# Create month identifier for grouping
df_monthly_prep['YearMonth'] = df_monthly_prep['Date'].dt.to_period('M')

# Aggregate by month
df_monthly = df_monthly_prep.groupby('YearMonth').agg({
    'Date': ['min', 'max'],
    'Month_Name': 'first',
    'Year': 'first',
    'Partials': 'sum',
    'Completed': 'sum',
    'Discovery Calls': 'sum',
    'Intro Calls': 'sum'
}).reset_index()

# Flatten column names
df_monthly.columns = ['YearMonth', 'Month_Start', 'Month_End', 'Month_Name', 'Year',
                      'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']

# Create Month label (e.g., "January 2025")
df_monthly['Month'] = df_monthly['Month_Name'] + ' ' + df_monthly['Year'].astype(str)

# Create Date Range
df_monthly['Date Range'] = (df_monthly['Month_Start'].dt.strftime('%m/%d/%Y') + ' - ' + 
                            df_monthly['Month_End'].dt.strftime('%m/%d/%Y'))

# Select and reorder columns
df_monthly = df_monthly[['Month', 'Date Range', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']]

print("✓ Monthly data aggregated successfully")
print(f"\nMonthly dataset shape: {df_monthly.shape[0]} rows × {df_monthly.shape[1]} columns")
print(f"\nAll months:")
display(df_monthly)
```

### Create Quarterly Aggregated Data

```python
# Work with daily data
df_quarterly_prep = df_daily.copy()

# Convert Date to datetime
df_quarterly_prep['Date'] = pd.to_datetime(df_quarterly_prep['Date'])

# Extract year and quarter
df_quarterly_prep['Year'] = df_quarterly_prep['Date'].dt.year
df_quarterly_prep['Quarter_Num'] = df_quarterly_prep['Date'].dt.quarter

# Create quarter identifier for grouping
df_quarterly_prep['YearQuarter'] = (df_quarterly_prep['Year'].astype(str) + 'Q' + 
                                    df_quarterly_prep['Quarter_Num'].astype(str))

# Aggregate by quarter
df_quarterly = df_quarterly_prep.groupby('YearQuarter').agg({
    'Date': ['min', 'max'],
    'Partials': 'sum',
    'Completed': 'sum',
    'Discovery Calls': 'sum',
    'Intro Calls': 'sum'
}).reset_index()

# Flatten column names
df_quarterly.columns = ['Quarter', 'Quarter_Start', 'Quarter_End',
                        'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']

# Create Date Range
df_quarterly['Date Range'] = (df_quarterly['Quarter_Start'].dt.strftime('%m/%d/%Y') + ' - ' + 
                              df_quarterly['Quarter_End'].dt.strftime('%m/%d/%Y'))

# Select and reorder columns
df_quarterly = df_quarterly[['Quarter', 'Date Range', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']]

print("✓ Quarterly data aggregated successfully")
print(f"\nQuarterly dataset shape: {df_quarterly.shape[0]} rows × {df_quarterly.shape[1]} columns")
print(f"\nAll quarters:")
display(df_quarterly)
```

### Create Annual Aggregated Data

```python
# Work with daily data
df_annual_prep = df_daily.copy()

# Convert Date to datetime
df_annual_prep['Date'] = pd.to_datetime(df_annual_prep['Date'])

# Extract year
df_annual_prep['Year'] = df_annual_prep['Date'].dt.year

# Aggregate by year
df_annual = df_annual_prep.groupby('Year').agg({
    'Date': ['min', 'max'],
    'Partials': 'sum',
    'Completed': 'sum',
    'Discovery Calls': 'sum',
    'Intro Calls': 'sum'
}).reset_index()

# Flatten column names
df_annual.columns = ['Year', 'Year_Start', 'Year_End',
                     'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']

# Create Date Range
df_annual['Date Range'] = (df_annual['Year_Start'].dt.strftime('%m/%d/%Y') + ' - ' + 
                           df_annual['Year_End'].dt.strftime('%m/%d/%Y'))

# Select and reorder columns
df_annual = df_annual[['Year', 'Date Range', 'Partials', 'Completed', 'Discovery Calls', 'Intro Calls']]

print("✓ Annual data aggregated successfully")
print(f"\nAnnual dataset shape: {df_annual.shape[0]} rows × {df_annual.shape[1]} columns")
print(f"\nAll years:")
display(df_annual)
```

## CUSTOM DATE RANGE ANALYSIS

### Load Custom Date Range Parameters

```python
# Load analysis reference dates from CSV
ref_dates_path = '../analysis_ref_date.csv'

# Check if file exists
if not os.path.exists(ref_dates_path):
    print(f"❌ Error: File not found at {ref_dates_path}")
    custom_start = None
    custom_end = None
else:
    # Load reference dates
    ref_dates = pd.read_csv(ref_dates_path, header=None)
    
    # Parse custom date range
    custom_start = None
    custom_end = None
    
    for idx, row in ref_dates.iterrows():
        if row[0] == 'Custom Start Date':
            try:
                custom_start = pd.to_datetime(row[1])
            except:
                custom_start = None
        elif row[0] == 'Custom End Date':
            try:
                custom_end = pd.to_datetime(row[1])
            except:
                custom_end = None
    
    print(f"Custom Date Range:")
    print(f"  Start: {custom_start if custom_start else 'Not specified'}")
    print(f"  End: {custom_end if custom_end else 'Not specified'}")
    
    if custom_start and custom_end:
        date_range_days = (custom_end - custom_start).days + 1
        print(f"  Duration: {date_range_days} days")
    else:
        print("\n⚠️ Custom date range not properly specified in analysis_ref_date.csv")
```

### Generate Custom Date Range Typeform Submissions Data

```python
print("\n" + "="*70)
print("CUSTOM DATE RANGE TYPEFORM SUBMISSIONS AGGREGATION (WITH PREVIOUS PERIOD)")
print("="*70 + "\n")

if custom_start and custom_end:
    # Calculate previous period dates - shift back by ONE MONTH, keeping same day-of-month
    from dateutil.relativedelta import relativedelta
    
    prev_start = custom_start - relativedelta(months=1)
    prev_end = custom_end - relativedelta(months=1)
    
    # Calculate duration for display
    date_range_duration = (custom_end - custom_start).days + 1
    
    print(f"Current Period (Requested): {custom_start.strftime('%m/%d/%Y')} - {custom_end.strftime('%m/%d/%Y')} ({date_range_duration} days)")
    print(f"Previous Period (Requested): {prev_start.strftime('%m/%d/%Y')} - {prev_end.strftime('%m/%d/%Y')} ({date_range_duration} days)\n")
    
    # Work with cleaned data for filtering
    df_custom_prep = df_cleaned.copy()
    
    # Convert Start Date to datetime
    df_custom_prep['AggregationDate'] = pd.to_datetime(df_custom_prep[start_date_col])
    
    # Filter data for CURRENT custom date range
    df_custom_filtered = df_custom_prep[
        (df_custom_prep['AggregationDate'] >= custom_start) & 
        (df_custom_prep['AggregationDate'] <= custom_end)
    ].copy()
    
    # Identify discovery and intro call columns (only once)
    discovery_col = None
    intro_col = None
    
    for col in df_custom_prep.columns:
        if 'discovery' in col.lower() and 'book' in col.lower():
            discovery_col = col
        if 'intro' in col.lower() and 'book' in col.lower():
            intro_col = col
    
    # Create empty list to store rows
    custom_rows = []
    
    # Process CURRENT period
    if len(df_custom_filtered) > 0:
        # Get ACTUAL start and end dates from the current period data
        actual_current_start = df_custom_filtered['AggregationDate'].min()
        actual_current_end = df_custom_filtered['AggregationDate'].max()
        
        # Count Response Types
        total_partials = (df_custom_filtered['Response Type'] == 'partial').sum()
        total_completed = (df_custom_filtered['Response Type'] == 'completed').sum()
        
        # Count Discovery and Intro Calls
        if discovery_col:
            total_discovery = (df_custom_filtered[discovery_col].notna() & 
                             (df_custom_filtered[discovery_col].astype(str).str.strip() != '')).sum()
        else:
            total_discovery = 0
        
        if intro_col:
            total_intro = (df_custom_filtered[intro_col].notna() & 
                         (df_custom_filtered[intro_col].astype(str).str.strip() != '')).sum()
        else:
            total_intro = 0
        
        # Use ACTUAL dates from current period data
        current_row = {
            'Date Range': f"{actual_current_start.strftime('%m/%d/%Y')} - {actual_current_end.strftime('%m/%d/%Y')}",
            'Partials': int(total_partials),
            'Completed': int(total_completed),
            'Discovery Calls': int(total_discovery),
            'Intro Calls': int(total_intro)
        }
        custom_rows.append(current_row)
        
        print(f"✅ Current period data found: {len(df_custom_filtered)} records")
        print(f"   Actual dates: {actual_current_start.strftime('%m/%d/%Y')} - {actual_current_end.strftime('%m/%d/%Y')}")
        print(f"   Total Partials: {total_partials}")
        print(f"   Total Completed: {total_completed}")
        print(f"   Total Discovery Calls: {total_discovery}")
        print(f"   Total Intro Calls: {total_intro}")
        
        # NOW calculate previous period based on ACTUAL current dates (shift by 1 month)
        actual_prev_start = actual_current_start - relativedelta(months=1)
        actual_prev_end = actual_current_end - relativedelta(months=1)
        
        # Filter data for PREVIOUS period using the shifted actual dates
        df_prev_filtered = df_custom_prep[
            (df_custom_prep['AggregationDate'] >= actual_prev_start) & 
            (df_custom_prep['AggregationDate'] <= actual_prev_end)
        ].copy()
        
        # Process PREVIOUS period
        if len(df_prev_filtered) > 0:
            # Count Response Types
            total_partials = (df_prev_filtered['Response Type'] == 'partial').sum()
            total_completed = (df_prev_filtered['Response Type'] == 'completed').sum()
            
            # Count Discovery and Intro Calls
            if discovery_col:
                total_discovery = (df_prev_filtered[discovery_col].notna() & 
                                 (df_prev_filtered[discovery_col].astype(str).str.strip() != '')).sum()
            else:
                total_discovery = 0
            
            if intro_col:
                total_intro = (df_prev_filtered[intro_col].notna() & 
                             (df_prev_filtered[intro_col].astype(str).str.strip() != '')).sum()
            else:
                total_intro = 0
            
            # Use the calculated previous period dates (shifted from actual current)
            prev_row = {
                'Date Range': f"{actual_prev_start.strftime('%m/%d/%Y')} - {actual_prev_end.strftime('%m/%d/%Y')}",
                'Partials': int(total_partials),
                'Completed': int(total_completed),
                'Discovery Calls': int(total_discovery),
                'Intro Calls': int(total_intro)
            }
            custom_rows.append(prev_row)
            
            print(f"\n✅ Previous period data found: {len(df_prev_filtered)} records")
            print(f"   Actual dates: {actual_prev_start.strftime('%m/%d/%Y')} - {actual_prev_end.strftime('%m/%d/%Y')}")
            print(f"   Total Partials: {total_partials}")
            print(f"   Total Completed: {total_completed}")
            print(f"   Total Discovery Calls: {total_discovery}")
            print(f"   Total Intro Calls: {total_intro}")
            
        else:
            print(f"\n⚠️ No data for previous period: {actual_prev_start.strftime('%m/%d/%Y')} - {actual_prev_end.strftime('%m/%d/%Y')}")
        
    else:
        print(f"⚠️ No data for current period: {custom_start.strftime('%m/%d/%Y')} - {custom_end.strftime('%m/%d/%Y')}")
    
    # Create final dataframe
    if len(custom_rows) > 0:
        df_custom = pd.DataFrame(custom_rows)
        
        # Display the results
        print("\n" + "="*70)
        print("Custom Date Range Typeform Submissions Data (with Previous Period):")
        print("="*70)
        display(df_custom)
        
    else:
        print("\n⚠️ No data available for either current or previous period")
        df_custom = None
        
else:
    print("❌ Cannot generate custom analysis - dates not properly specified")
    print("   Please check Custom Start Date and Custom End Date in analysis_ref_date.csv")
    df_custom = None

print("\n" + "="*70)
```
